# Record validation practice
Before running, predict which of the three fictional records passes and explain why. Record your prediction in Student.


```python
from jsonschema import Draft202012Validator
schema = {"type": "object", "required": ["name", "age"], "properties": {"name": {"type": "string"}, "age": {"type": "integer", "minimum": 0}}}
validator = Draft202012Validator(schema)
```

```python
records = [{"name": "Ada", "age": 37}, {"name": "Lee", "age": "37"}, {"age": 37}]
observed = [validator.is_valid(record) for record in records]
print("Observed validation:", observed)
assert observed == [True, False, False]
```
